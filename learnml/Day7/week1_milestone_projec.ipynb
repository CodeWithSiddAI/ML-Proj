```python
import numpy as np

# 1. THE DATASET
# X = [Size (sqft), Bedrooms, Age (years)]
X_train = np.array([
    [2104, 5, 1], 
    [1416, 3, 20], 
    [1534, 3, 30], 
    [852, 2, 36]
])
# y = Price in thousands of dollars ($460k, $232k, etc.)
y_train = np.array([460, 232, 315, 178])

# 2. THE ENGINE
def compute_cost(X, y, w, b):
    m = X.shape[0] # Number of houses
    
    # Fully Vectorized: X (4x3 matrix) dot w (3-item vector) gives all 4 predictions instantly!
    y_hat = np.dot(X, w) + b
    
    error = y_hat - y
    cost = (1/(2*m)) * np.sum(error**2)
    
    return cost

def compute_gradient(X, y, w, b):
    m = X.shape[0]
    
    # Calculate predictions and error vector
    y_hat = np.dot(X, w) + b
    error = y_hat - y
    
    # Fully Vectorized Gradients! 
    # X.T flips the matrix so the math lines up perfectly. 
    # No 'for' loops needed to calculate all 3 weights.
    dj_dw = (1/m) * np.dot(X.T, error)
    dj_db = (1/m) * np.sum(error)
    
    return dj_dw, dj_db

def gradient_descent(X, y, w_in, b_in, alpha, epochs):
    w = w_in
    b = b_in
    
    for i in range(epochs):
        dj_dw, dj_db = compute_gradient(X, y, w, b)
        
        # Update w and b using the learning rate (alpha)
        w = w - (alpha * dj_dw)
        b = b - (alpha * dj_db)
        
        # Print progress
        if i % 100 == 0:
            cost = compute_cost(X, y, w, b)
            # Printing bias to 4 decimal places so you can see it's moving microscopically!
            print(f"Epoch {i:4}: Cost = {cost:8.2f} | Bias = {b:8.4f}")
            
    return w, b

# 3. RUN THE PIPELINE
# Initialize w with zeros (one zero for each of the 3 features)
initial_w = np.zeros(X_train.shape[1]) 
initial_b = 0.

# Notice how tiny the learning rate is! Multiple features make the gradients huge.
iterations = 1000
alpha = 5.0e-7 

print("Starting Training...")
final_w, final_b = gradient_descent(X_train, y_train, initial_w, initial_b, alpha, iterations)

print("\n--- Training Complete ---")
print(f"Final Weights: {final_w}")
print(f"Final Bias: {final_b:.2f}")

# Let's predict the first house in our dataset just to see!
first_house_prediction = np.dot(X_train[0], final_w) + final_b
print(f"\nPrediction for House 1: ${first_house_prediction * 1000:,.2f}")
print(f"Actual Price of House 1: ${y_train[0] * 1000:,.2f}")
```
