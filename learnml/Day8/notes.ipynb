### Lecture 1: The "Yes or No" Problem

Imagine you are building a Machine Learning model for a hospital. You measure the size of a tumor ($x$), and you need to predict if it is Malignant (Dangerous) or Benign (Safe).

This is a Yes/No question. In Machine Learning, we represent this with numbers:

* **0** = Negative Class (Benign / Safe)
* **1** = Positive Class (Malignant / Dangerous)

**Why can't we just use our Week 1 Linear Regression line for this?**
Let's say you train your Week 1 model to draw a straight line through the data. You tell the computer: *"If the line outputs a number greater than 0.5, predict Malignant (1). If it's less than 0.5, predict Benign (0)."*

This works okay at first. But then, a patient comes in with a *massively* huge tumor that is Malignant. Because Linear Regression tries to keep the line close to *all* data points, this giant outlier will drag the entire slope of your line upward. Suddenly, your 0.5 threshold shifts, and you start misdiagnosing healthy patients as sick!

Furthermore, Linear Regression is just a straight line that goes on forever. If a tumor is huge, your Linear Regression model might predict `y = 4.7`. What does a 4.7 mean when the only options are 0 or 1? It makes no mathematical sense.

### Lecture 2: The Decision Boundary

To solve this, we shift our entire way of thinking.

In Week 1, we drew a line **through** the dots to predict their value.
In Week 2, we draw a line **between** the dots to separate them into groups.

This line of separation is called a **Decision Boundary**.

* Everything on the left side of the boundary is classified as 0.
* Everything on the right side of the boundary is classified as 1.

The algorithm's new job isn't to guess a price; its job is to figure out exactly where to draw that fence.

### Lecture 1: The Score ($z$) and the Threshold

In Week 1, our equation $\hat{y} = \mathbf{w} \cdot \mathbf{x} + b$ gave us a house price.
In Week 2, we use the exact same equation, but we stop calling the answer $\hat{y}$. Instead, we call it **$z$ (The Score)**.

$$z = (w_1x_1) + (w_2x_2) + b$$

Think of $z$ as a **Balance Scale**.

* If $z$ is a **positive number** (greater than 0), the scale tips to the right. The computer screams: **"Class 1! (Spam / Malignant)"**
* If $z$ is a **negative number** (less than 0), the scale tips to the left. The computer screams: **"Class 0! (Not Spam / Benign)"**
* If $z$ is **exactly 0**, the scale is perfectly balanced. The computer is 50/50 unsure.

This creates a massive realization: The exact location where $z = 0$ is the **Decision Boundary**. It is the invisible fence separating the two classes.

### Lecture 2: The Geometry of the Boundary

Let’s look at your Spam Email scenario mathematically.

* $x_1$ = Number of spelling mistakes.
* $x_2$ = Number of "URGENT"s.

Imagine your machine learning algorithm has finished training, and it found these exact weights:

* $w_1 = 2$ (Each mistake adds 2 points of "Spamminess")
* $w_2 = 5$ (Each "URGENT" adds 5 points of "Spamminess")
* $b = -15$ (The base score. Notice it's negative? An email starts innocent, at -15, and has to *earn* its way into being marked as Spam).

So, our scoring equation for any email is:


$$z = 2x_1 + 5x_2 - 15$$

If we want to find the exact boundary fence, we set $z$ to 0:


$$0 = 2x_1 + 5x_2 - 15$$

This is just the algebraic equation for a line! If you rearrange it to solve for $x_2$ (just like $y = mx + b$ from high school), you can draw this exact line on a 2D graph. Everything above the line is Spam. Everything below is Not Spam.

### Lecture 3: The Flaw of the Hard Threshold

Today, we are using a "Hard Threshold" (if $z \ge 0$, it's Spam).
But imagine Email X gets a score of $z = 0.1$, and Email Y gets a score of $z = 10,000$.
With a Hard Threshold, the computer just says "Both are Spam (1)."

But as a human, you know Email Y is *definitely* spam, while Email X is right on the fence. We want the computer to tell us its **confidence** (e.g., "I am 99% sure Email Y is spam, but only 51% sure about Email X").

That leads us to tomorrow (Day 9), where we will learn the **Sigmoid Function**—a mathematical hydraulic press that takes this raw $z$ score and squishes it into a clean probability percentage between 0% and 100%.
