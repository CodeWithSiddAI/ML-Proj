### Lecture 1: The "And" Problem (Math)

Until today, our prediction math was simple:


$$\hat{y} = wx + b$$

But now, our $x$ (Feature) is no longer a single number. It is a list of features:

* $x_1$ = Size (e.g., 2000 sq ft)
* $x_2$ = Bedrooms (e.g., 4)
* $x_3$ = Age (e.g., 10 years old)

Because we have 3 features, a single weight ($w$) is no longer enough. We need a specific weight for *each* feature to represent how important it is.

* $w_1$ = How much does each square foot cost?
* $w_2$ = How much value does an extra bedroom add?
* $w_3$ = How much value does the house lose each year it gets older?

Our new, expanded equation looks like this:


$$\hat{y} = (w_1x_1) + (w_2x_2) + (w_3x_3) + b$$

### Lecture 2: Vector Notation (The Magic Trick)

If we had a dataset with 1,000 features (like 1,000 pixels in an image), writing out $w_1x_1 + w_2x_2 ... + w_{1000}x_{1000}$ would be a nightmare.

This is where the **Linear Algebra** we learned on Day 2 saves us.
We pack all our weights into a single Vector called $\mathbf{w}$.
We pack all our features into a single Vector called $\mathbf{x}$.

* $\mathbf{w} = [w_1, w_2, w_3]$
* $\mathbf{x} = [x_1, x_2, x_3]$

Instead of writing out that massive addition equation, we use the **Dot Product** (multiply the matching pairs, then add them all together).

In mathematical notation, we write the entire multi-dimensional prediction equation as elegantly as this:

$$\hat{y} = \mathbf{w} \cdot \mathbf{x} + b$$

*(Note: In textbooks, vectors are usually written in bold, or with a little arrow over them to remind you they are a list, not a single number).*

### Lecture 3: Vectorization (The "Stamping Machine")

Why do we care about packing things into Vectors? Because of **Speed**.

Imagine you work at a post office and you have to stamp 1,000 envelopes.

* **A Python `for` loop** is like a human. You pick up envelope 1, stamp it. Pick up envelope 2, stamp it. It takes time.
* **Vectorization (NumPy)** is like a giant hydraulic press. You lay all 1,000 envelopes on a table, pull a lever, and the machine stamps all 1,000 at the exact same millisecond.

Under the hood, NumPy doesn't use Python to do the math. It sends the vectors directly to your computer's CPU and uses C-code to calculate them all simultaneously. This is called **Vectorization**. Andrew Ng stresses that *you should never use a `for` loop in Machine Learning if a vectorized NumPy operation can do it instead.*

```python
import numpy as np

# 1. The Data (Vectors)
x = np.array([2000, 4, 10])
w = np.array([150, 10000, -2000])
b = 50000

print(f"--- Predicting House Price ---")

# ==========================================
# METHOD 1: The Human Way (For Loop)
# ==========================================
prediction_loop = 0 
# YOUR CODE HERE (Hint: loop through the items in x and w, multiply them, add to prediction_loop. Don't forget to add 'b' at the end!)
total = 0
for i in range(len(x)):
    total += x[i] * w[i]

prediction_loop = total + b 




print(f"Prediction using For Loop:      ${prediction_loop:,.2f}")

# ==========================================
# METHOD 2: The Machine Way (Vectorized)
# ==========================================
prediction_vectorized = 0
# YOUR CODE HERE (Hint: This should literally be one short line of code)
prediction_vectorized = np.dot(x, w) + b



print(f"Prediction using Vectorization: ${prediction_vectorized:,.2f}")
```
